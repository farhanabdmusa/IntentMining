```python
import pandas as pd
from google_play_scraper import Sort, reviews_all

# 1. Konfigurasi Scraping
app_id = 'com.ss.android.ugc.trill' # TIKTOK

print(f"Memulai pengambilan data review untuk: {app_id}...")

# 2. Ambil Semua Review (atau gunakan reviews() untuk membatasi jumlah)
scraped_data = reviews_all(
    app_id,
    sleep_milliseconds=1000, # Delay agar tidak diblokir Google
    lang='id',              # Bahasa Indonesia
    country='id',           # Region Indonesia
    sort=Sort.NEWEST,        # Ambil dari yang terbaru
    count=500
)

# 3. Ubah ke format Pandas DataFrame
df = pd.DataFrame(scraped_data)

# 4. Ambil kolom yang penting saja untuk Intent Mining
# Biasanya kita butuh 'content' untuk teks dan 'score' sebagai konteks tambahan
df_clean = df[['userName', 'score', 'content', 'at']]

# 5. Simpan ke CSV
df_clean.to_csv('review_aplikasi.csv', index=False)

print(f"Berhasil mengambil {len(df_clean)} review!")
print(df_clean.head())
```

```python
reviews_list = df_clean['content'].values.tolist()
```

```python
from ITER_DBSCAN import ITER_DBSCAN
```

```python
%%time
model = ITER_DBSCAN(initial_distance=0.3, initial_minimum_samples=16, delta_distance=0.01, delta_minimum_samples=1, max_iteration=15)
```

```python
%%time
labels = model.fit_predict(reviews_list)
```

```python
df['cluster_ids'] = labels
```
