```python
%%time
model = ITER_DBSCAN(
    initial_distance= 0.3,
    initial_minimum_samples=30, 
    delta_distance=0.01, 
    delta_minimum_samples=1, 
    max_iteration=15,
    threshold= 1000,
    embedding_model= "IndoSBERT",
)
```

```python
labels, evaluate_results = model.fit_predict_evaluate(filename=filepath, 
                                   filetype='csv', 
                                   text_column='content', 
                                   target_column='intent')
```
