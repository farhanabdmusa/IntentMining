```python
import pandas as pd

def mapping_intent(text):
    text = str(text).lower()
    
    # 1. Technical Bug/Crash
    if any(word in text for word in ['bug', 'keluar sendiri', 'error', 'elor', 'force close', 'fore close', 'mati sendiri', 'kembali sendiri', 'eroro']):
        return "Technical Bug/Crash"
    
    # 2. Account Issue
    elif any(word in text for word in ['akun', 'ben', 'banned', 'blokir', 'block', 'login', 'daftar', 'masuk', 'verifikasi']):
        return "Account Issue"
    
    # 3. Performance Issue
    elif any(word in text for word in ['lag', 'ngeleg', 'lemot', 'berat', 'panas', 'patah', 'loading', 'lelet', 'habis paket', 'kuota']):
        return "Performance Issue"
    
    # 4. Feature Complaint/Request
    elif any(word in text for word in ['fitur', 'toko', 'shop', 'keranjang', 'repost', 'draft', 'komentar', 'komen', 'scrol', 'update']):
        return "Feature Complaint/Request"
    
    # 5. Monetization/Earning
    elif any(word in text for word in ['uang', 'dana', 'saldo', 'hadiah', 'potongan harga', 'event', 'refund', 'koin']):
        return "Monetization/Earning"
    
    # 6. Praise & Gratitude
    elif any(word in text for word in ['bagus', 'suka', 'mantap', 'asik', 'seru', 'well', 'senang', 'terimakasih', 'kasih bintang', '😁', '🥰']):
        return "Praise & Gratitude"
    
    # 7. General Request/Complaint
    elif any(word in text for word in ['tolong', 'perbaiki', 'jelek', 'sampah', 'kecewa', 'parah', 'gajelas']):
        return "General Request/Complaint"
    
    return "Unlabeled/Noise"

# Memuat dataset
df = pd.read_csv('review_aplikasi.csv')

# Menerapkan mapping
df['intent'] = df['content'].apply(mapping_intent)

# Hapus Unlabeled/Noise intent
df = df.loc[df.intent != 'Unlabeled/Noise']

# Contoh cara cek kesesuaian cluster (Jika Anda sudah punya kolom 'cluster_id')
# comparison = df.groupby(['cluster_id', 'label_acuan']).size().reset_index(name='count')
# print(comparison)

# Simpan hasil untuk dicek manual
df.to_csv('review_dengan_intent.csv', index=False)

print("Mapping selesai! File 'review_dengan_inten.csv' telah dibuat.")
print("\nRingkasan Label Intent:")
print(df['intent'].value_counts())
```
